```python
import pandas as pd
import joblib

# Load dataset
df = pd.read_csv("../data/Breast-Cancer_data.csv")

# Rename the 3 columns to match the trained pipeline
df = df.rename(columns={
    "concave points_mean": "concave_points_mean",
    "concave points_se": "concave_points_se",
    "concave points_worst": "concave_points_worst"
})

# Load saved pipeline
pipeline = joblib.load("../models/final_svm_pipeline.pkl")

# Get exact model features
feature_names = pipeline.feature_names_in_

# Take first real patient
sample = df.loc[0, feature_names]

print("Actual diagnosis:", df.loc[0, "diagnosis"])

print("\nCOPY THIS JSON INTO SWAGGER:")
print(sample.to_json(indent=2))
```

```python
# Create a copy of the pipeline's expected feature names
feature_names = list(pipeline.feature_names_in_)

# Convert pipeline feature names to match current dataframe column names
feature_names = [
    name.replace("concave points", "concave_points")
    for name in feature_names
]

# Take first real patient
sample = df.loc[0, feature_names]

print("Actual diagnosis:", df.loc[0, "diagnosis"])

print("\nCOPY THIS JSON INTO SWAGGER:")
print(sample.to_json(indent=2))
```

```python
# Find the first benign patient
benign_index = df[df["diagnosis"] == "B"].index[0]

# Feature names compatible with the current dataframe
feature_names = [
    name.replace("concave points", "concave_points")
    for name in pipeline.feature_names_in_
]

# Select benign patient
benign_sample = df.loc[benign_index, feature_names]

print("Row index:", benign_index)
print("Actual diagnosis:", df.loc[benign_index, "diagnosis"])

print("\nCOPY THIS JSON INTO SWAGGER:")
print(benign_sample.to_json(indent=2))
```
